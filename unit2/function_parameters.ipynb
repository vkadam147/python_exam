actual parameter:
   1. the parameter which are pass while calling the function
   2.we can pass variable,literal,expression in a actual parameter

formal parameter:
    the parameter which are defined inside the function defination

```python
def add(a,b):#formal parameter
    return a+b
add(110,20)#actual parameter
```

1.Positional Parametres:
    1.These are the parameters passed to function while calling the function in correct position order
    2.The no of actual and formal parameters must be same
    3.The position of actual and formal parameters must be same else we will get wrong result


```python
def person(name,age): 
    print(f"Name:{name} age:{age}")

person("Vaishnavi",23)
person("Akshada",20)
person(24,"Shreya") # Here position for name and age is mismatched so   name is assigned as 24 and age is aasigned as shreya so we will get wrong result

```

2.Keyword Parameters:
    1.In Function call statement if we are passing parameters using the parameter name from formal parameters then that paramaters are known as keyword parametrs
    2.The name of formal paramaters and passed in actual paramaters must be matched in keyword paramaters
    3.We can use both positional and keyword paramaters simultaneously but first we have to take positional parameter then keyword parameter always else we will get syntax error
    4.If we are take keyword parameter first then position parameter then we will get syntax error 

```python
def greet(name,message):
    print(f"Name:{name} Message:{message}")

greet("Manavi","Hello")
greet(message="Good Morning",name="Vaishu")
# greet("Good Afternoon",name="Sanskriti") 
# Both sanskriti and good afternoon are directly assigned to name as keyword parameter name assigned as sanskriti and as first parameter is name and position parameter is also first so good afternoon is also directly assigning to name so we got multiple parameters assigned to single parameter name
# greet(message="Good Night","alexa")  ===>SyntaxError as we have taken first keyword argumenet then position parameter so it will  get syntax error
```

3.Default Parameter:
    1.if we are assigning default values for formal parameters while declaring function that parameter is known as default parameter
    2.if we are created function using default parameter then that parameter is optional if we  do not pass that parameter while calling the function then python will take default value for that parameter that we have assigned while declaring the function
    3.if we are passing the value then  default value will be overrided as passed value will be used while function execution
    4.default parameter must be after non default parameter

```python
def greet(name='Alexa'):
    print(f"Hello {name}")

greet("Vaishu")
greet()

def add(a,b=10):
    print(a+b)
add(10,20)
add(18) #a=18 b=10
add(24) # a=24 b=10
```

4.Variable length argument(*args):
    1.if we want to pass n no of arguement to function then we should go for variable length argument
    2.we can declare variable length parameter with * symbol (*vararg) 
    3.*vararg act as variable length argument and data in variable length argument is stored as tuple
    4.type(*vararg)=tuple
    5.if we want to take any arguement after variable length argument then taht argument must be keyword argument else we will get syntax error 

```python
def m1(*vaishu):
    print(type(vaishu))
    print(f"length of tuple vaishu is :{len(vaishu)}")
    for x in vaishu:
        print(x)
    print("End")
    print("========================================================")
m1()
m1(10)
m1(10,20)
m1(10,20,30)
m1(10,20,30,40)
```

5.Keyword variable length argument:
    1.If we want to pass any no of 'keyword argument' then we should use keyword variable length argument
    2.we can decalare keword variable length argument using "**paramater_name"
    3.keyword variable  length arguement is stored as dictionary
    4.we can pass 0 to n no of keyword argument
    5.keyword argument must be at the end of parameter
    6.order of keyword arguemnet does not matter as it is stored as dictionary and dictionary is unordered


```python
def show(**a):
    if len(a)==0:
        print("Dictionary is empty")
    print(f"Type of a is {type(a)}")
    for key,value in a.items():
        print(key,value)
    print("--------------------------------------------------------------")

show(name='Vaishnavi',course='MCA')
show(name='Akshada',course='Diploma')
show(name='Sanskriti',course='MCA')
show(a=10,b=20,c=30,d=40,e=50,f=60)
show()
```
