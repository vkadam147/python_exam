generator function:
    1.generator is type of function which is used to return one value at a time by pausing the current state of function using yield keyword
    2.generator function does not return all result at a time
    3.genrator function result one by one by pausing the current state of function
    4.gennerator function used yield keyword for returning the value


what is yield:
    1.yield is keyword which is used to return value and pause the  execution while saving the  current state of function
        return: Terminates the execution of function
        yield: Pauses the execution of function

what is next():
    1.next() is built in function that takes generator object as input and return yielded value which is coming from generator
    2.if generator does not have any value then next() function will raise StopIteration exception
    3.next() function will execute the function code untill the next yield keyword
    4.after yielding value it pauses and saves cuurent state of function

```python
def my_genrator():
    a=0
    for i in range(5):
        print(i,a)
        a+=1
        
    yield "first yield"
    print("seond next() started")

    print("---------------------------------")

    for i in range(5,10):
        print(i,a)
        a+=1
    yield "second yield"
    print("Third next() started")

    print("---------------------------------")
    for i in range(10,15):
        print(i,a)
        a+=1

    yield "Third yield"
    print("---------------------------------")

gne=my_genrator()
print(gne)
print(next(gne))
# print(next(gne))
# print(next(gne))
# print(next(gne))


```
