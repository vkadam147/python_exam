scope of variable:  
        1.it defines accessibility and visibillity of the variable 
        2.accessibility means where we can access variable and where we cant access
        3.visibility means:where the variable is visible
        THERE ARE 4 SCOPE OF VARIABLE DEFINED AS LEGB RULE
            1.L:=Local scope
            2.E:=Enclosing scope
            3.G:=Global scope
            4.B:=Buit in scope

1.Local scope:
        1.variable declare inside function have local scope
        2.local variable are stored inside heap and refenrence are stored inside local namespace
        3.as soon as function execution ends local variable is no longer available
        4.local scope has a highest priority
        


```python
def m1():
    a=20 #variable a has local scope
    print(a)

m1()
print(a)#we can not accessed local variable a outside function because a is local to m1
```

2.Enclosing
        1.enclosing scope comes in nested function 
        2.if there are two function which are nested:
            def outer():
                a=10
                def inner():
                    print(a)
        3.the variable inside outer function (a) have enclosing scope when it is accessed inside inner function
        4.inside inner function enclosing scope variable have read only authority
        5.if we tried to update enclosing scope variabkle inside inner function then local variable will be created for inner function
        6.we can not modify enclosing scope variable inside inner function directly 
        7.to update enclosing scope variable inside inner function we have to use nonlocal keyword
        8.if we use nonlocal keyword inside inner function then we can modify enclosing scope variable directly

```python
def outer():
    a=10
    def inner():
        print(f"inner : address of a:{id(a)},A:{a}")
    print(f"outer: address of a:{id(a)},A:{a}")
    inner()
outer()
    
    
       
```

Global scope:
        1.the variable or function or class declared outside function,class have global scope
        2.the variable declared outside function or class are known as global variable
        3.global scope members reference stored inside global namespace and actual object created inside heap
        4.global members are available untill the end of the program 
        5.we can not modify global members inside local scope/enclosing scope
        6.global variables are read only for local/enclosing scope
        7.if we want to modify global variable inside local/enclosing  scope then we have to use global keyword
        8.if we declare variable inside local/enclosing with global keyword then we can directly modify inside local/enclosing scope
