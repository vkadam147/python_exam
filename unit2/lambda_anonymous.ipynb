Anonynous Fuction:
    1.The Function Wiithout name and created using lambda keyword is known as anonymous function
    2.Anonymous function is best for short and simple operations
    3.We can create anonymous function suing lambda keyword
    4.Syntax:
        lambda arguments:expression
        1.lambda: It is used to create anonynous function
        2.argumenets: The argumenet which is required for function (0 to n)
        3.expression: single line operation to perform task


```python
square=lambda a:a**2
print(square(5))
print(square)
```

```python
addition=lambda a,b:a+b
print(addition(2,2))
```

```python
sum_tuple=lambda a:sum(a)
print(sum_tuple([1,2,3,4]))
```

```python
power1=lambda base,power=2:base**power
print(power1(2,3))
print(power1(8))
```

```python
greet=lambda:"good morning"
print(greet())
```

```python
max_num=lambda a,b:a if a>b else b
print(max_num(2,3))
print(max_num(3,2))
```

```python
length=lambda a:len(a)
print(length("monu"))
```

```python
# map() using lambda fucntion
l=[1,2,3,4]
squares=list(map(lambda a:a**2,l))
print(squares)
```

```python
#map
l=[1,2,3,4]
square=list(map(lambda a:a**2,l))
print(square)
```

```python
#using map with user define fun
def sq(n):
    return n**2

nums=[1,2,3,4]
square=list(map(sq,nums))
print(square)
```

```python
a=["vaishnavi","shreya","akshada"]

c=list(map(lambda x:x.upper(),a))
print(c)
```

```python
#map with multiple iterable
l=[1,2,3]
l2=[4,5,6]

add_two_list=list(map(lambda x,y:x+y,l,l2))
print(add_two_list)
```

```python
#converting data types using map()
'''string list into int list'''

data=['1','2','3','4']
result=list(map(int,data))
print(result)
```
