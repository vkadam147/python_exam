1.Function is Block of code which is used to perform specific task when it is called
2.Function may have parameters or may not have any parameters(Parameters are optional)
3.Function may return or may not return anything (return is optional)
4.Function is created using def keyword
5.It will execute only if we called explicitely

why use function?
 1.To increase code readability
 2.To increase modularity of programming
 3.It helps while debugging and Testing
 4.To reduce length of code


syntax:
    def function_name(parameters):
        statements
        return value

Note:parameters and return is optional

There are two types of parameters:
1.Actual parameter
2.Formal parameter

1.actual parameter
the parameter which are pass while calling the function 

2.formal parameter:
the parameter which are declared inside the function defination

```python
def outer():
    num=100
    def inner():
        nonlocal num
        num=200
        print(f"inner id :{id(num)},num:{num}")
    print(f"outer id:{id(num)},num:{num}")
    inner()
    print(f"outer id:{id(num)},num:{num}")
outer()
```
