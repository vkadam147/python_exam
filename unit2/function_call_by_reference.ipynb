```python
def modify(a):
    print(f"address of a: {id(a)}")
    print(f"A:{a}")
    a.append(5)
    print(f"address after update:{id(a)}")
    print(f"A:{a}")

name=[1,2,4,6,7]
print(f"address of name:{id(name)}")
print(f"name:{name}")

modify(name)
print(f"address after modify:{id(name)}")
print(f"name after modify():{name}")
```
